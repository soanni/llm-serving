```python
from transformers import AutoModelForCausalLM, AutoTokenizer
```

```python
import torch
```

```python
from bertviz import head_view, model_view
```

```python
path = "/srv/models/qwen2.5-0.5B"
```

```python
# model = AutoModelForCausalLM.from_pretrained(path, attn_implementation="eager").eval().cuda()
model = AutoModelForCausalLM.from_pretrained(path, output_attentions=True).eval().cuda()
```

```python
text = "write a short introduction about the US capital"
```

```python
tokenizer = AutoTokenizer.from_pretrained(path, output_attentions=True)
```

```python
inputs = tokenizer(text, return_tensors="pt").to(model.device)
inputs
```

```python
with torch.no_grad():
    outputs = model(**inputs, output_attentions=True)
```

```python
attention = outputs.attentions
#attention
```

```python
tokens = tokenizer.convert_ids_to_tokens(inputs['input_ids'][0])
# tokens = [t.replace('Ġ', '▁') for t in tokens]
tokens
```
