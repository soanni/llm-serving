```python
from transformers import pipeline, AutoModelForCausalLM, AutoTokenizer
import time, torch, gc
```

```python
def free_gpu(model):
    if model:
        del model
    if torch.cuda.is_available:
        torch.cuda.empty_cache()
        torch.cuda.reset_peak_memory_stats()
    gc.collect()
```

```python
model_path = "/srv/models/qwen2.5-0.5B"
```

```python
start_time = time.time()
```

```python
tokenizer = AutoTokenizer.from_pretrained(model_path, trust_remote_code=True)
model = AutoModelForCausalLM.from_pretrained(model_path, trust_remote_code=True).to("cuda" if torch.cuda.is_available() else "cpu")
```

```python
prompt = "United States citizens have a bunch of benefits like"
```

```python
generator = pipeline("text-generation", model=model_path, tokenizer=tokenizer)
```

```python
outputs = generator(prompt, max_new_tokens=128, temperature=0.8, top_p=0.95)
```

```python
end_time = time.time()
```

```python
for out in outputs:
    print(f"Generated text: {out}")
    print(f"Time taken: {end_time - start_time:.2f} seconds")
```
