```python
from transformers import AutoModelForCausalLM
```

```python
path = "/srv/models/qwen2.5-0.5B"
```

```python
model = AutoModelForCausalLM.from_pretrained(
    path,
    device_map="cuda",
    trust_remote_code=True
)
```

```python
config = model.config
```

```python
print("\n === Model configuration parameters ===")
```

```python
print("\nArchitecture parameters\n")
print(f"Hidden size: {config.hidden_size}")
print(f"Number of layers: {config.num_hidden_layers}")
print(f"Number of attention heads: {config.num_attention_heads}")
print(f"Intermidiate size: {config.intermediate_size}")
print("\nTokenizer parameters\n")
print(f"Vocabulary size: {config.vocab_size}")
print(f"Maximum position embeddings: {config.max_position_embeddings}")
print("\nModel-specific parameters:\n")
for k,v in config.to_dict().items():
    print(f"{k}: {v}")
```
