```python
from transformers import pipeline
```

```python
model_path = "/srv/models/qwen2.5-0.5B"
```

```python
generator = pipeline("text-generation", model_path, device="cuda")
```

```python
prompt = "Write a short introduction about the US capital city."
```

```python
generatated_text = generator(prompt, max_length=50, num_return_sequences=1)
```
