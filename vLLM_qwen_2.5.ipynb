```python
from vllm import LLM, SamplingParams
```

```python
import time, torch, gc
```

```python
def free_gpu(model):
    if model:
        del model
    if torch.cuda.is_available:
        torch.cuda.empty_cache()
        torch.cuda.reset_peak_memory_stats()
    gc.collect()
```

```python
model_path = "/srv/models/qwen2.5-0.5B"
```

```python
# prompt = "The history of human … … … How might the next wave of communication tools shape our relationships, societies, and sense of identity?"
prompt = "United States citizens have a bunch of benefits like"
```

```python
llm = LLM(model=model_path, dtype="float16")
```

```python
sampling_params = SamplingParams(temperature=0.8, top_p=0.95, max_tokens=128)
```

```python
start_time = time.time()
outputs = llm.generate([prompt], sampling_params)
end_time = time.time()
for out in outputs:
    print(f"Generated text: {out}")
    print(f"Time taken is {end_time - start_time:.2f} seconds")
free_gpu(llm)
```
